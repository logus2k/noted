```python
# Training dynamics visualization
import matplotlib.pyplot as plt
import numpy as np

# Training analytics dashboard
fig, ax = plt.subplots(1, 3, figsize=(14, 4))

# Loss curves
e = np.arange(1, 31)
ax[0].plot(e, 0.95 * np.exp(-0.12 * e) + 0.08 + np.random.normal(0, 0.005, 30), label='Train', color='#4a90d9', lw=2)
ax[0].plot(e, 0.90 * np.exp(-0.10 * e) + 0.10 + np.random.normal(0, 0.008, 30), label='Val', color='#e67e22', lw=2)
ax[0].axvline(x=21, color='#c0392b', ls='--', alpha=0.5, label='Early Stop')
ax[0].set(xlabel='Epoch', ylabel='Loss', title='Training Convergence')
ax[0].legend(fontsize=9); ax[0].grid(True, alpha=0.3)

# Prediction scatter
actual = np.random.normal(10, 5, 200)
ax[1].scatter(actual, actual + np.random.normal(0, 1.5, 200), alpha=0.4, s=15, color='#4a90d9')
ax[1].plot([actual.min() - 1, actual.max() + 1], [actual.min() - 1, actual.max() + 1], '--', color='#c0392b', lw=1.5)
ax[1].set(xlabel='Actual (C)', ylabel='Predicted (C)', title='Predictions vs Actual')
ax[1].grid(True, alpha=0.3)

# Feature importance
feat, imp = ['T (C)','p (mbar)','rh (%)','wv (m/s)','wd (deg)','rain'], [.35,.22,.18,.12,.08,.05]
ax[2].barh(feat, imp, color=['#4a90d9','#5ba3e8','#7bb8f0','#9ccdf5','#bde0fa','#dff0ff'])
ax[2].set(xlabel='Importance', title='Feature Importance'); ax[2].invert_yaxis(); ax[2].grid(True, alpha=0.3, axis='x')

plt.suptitle('MLOps Dashboard - Training Analytics', fontsize=14, y=1.02)
plt.tight_layout(); plt.show()
```

```python
# Interactive time series with Plotly
import plotly.graph_objects as go

np.random.seed(42)
days = 365
t = np.arange(days)
trend = 0.02 * t
seasonal = 5 * np.sin(2 * np.pi * t / 365) + 2 * np.sin(2 * np.pi * t / 7)
noise = np.random.normal(0, 1.5, days)
temperature = 12 + trend + seasonal + noise

fig = go.Figure()
fig.add_trace(go.Scatter(x=t, y=temperature, mode='lines',
    name='Temperature', line={'color': '#4a90d9', 'width': 1.5}))
fig.add_trace(go.Scatter(x=t, y=12 + trend + seasonal, mode='lines',
    name='Trend + Seasonal', line={'color': '#e67e22', 'width': 2, 'dash': 'dash'}))
fig.update_layout(
    title='Daily Temperature Forecast - Synthetic Data',
    xaxis_title='Day', yaxis_title='Temperature (C)',
    template='plotly_white', height=400,
    legend={'x': 0.01, 'y': 0.99, 'bgcolor': 'rgba(255,255,255,0.8)'})
fig.show()
```

```python
# Model comparison heatmap
import matplotlib.pyplot as plt
import seaborn as sns

models = ['GRU', 'LSTM', 'Linear', 'XGBoost', 'Prophet']
metrics = ['MAE', 'RMSE', 'R2', 'Train Time (s)', 'Parameters']
data = np.array([
    [2.06, 2.59, 0.89, 84, 92952],
    [2.12, 2.65, 0.88, 91, 98304],
    [3.41, 4.12, 0.72, 0.3, 12],
    [2.31, 2.87, 0.86, 12, 450],
    [2.78, 3.42, 0.81, 8, 0],
])

# Normalize each column to 0-1 for heatmap
norm = (data - data.min(axis=0)) / (data.max(axis=0) - data.min(axis=0) + 1e-8)

fig, ax = plt.subplots(figsize=(10, 4))
sns.heatmap(norm, annot=data, fmt='.2f', xticklabels=metrics, yticklabels=models,
    cmap='RdYlGn_r', linewidths=0.5, ax=ax, cbar_kws={'label': 'Relative (lower = better)'})
ax.set_title('Model Comparison - Jena Weather Forecasting', fontsize=14, pad=12)
plt.tight_layout()
plt.show()
```
