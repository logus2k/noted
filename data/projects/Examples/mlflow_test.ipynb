```python
from sklearn.linear_model import LinearRegression
import numpy as np

X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

model = LinearRegression()
model.fit(X, y)
print(f"Score: {model.score(X, y)}")
print(f"Coef: {model.coef_}, Intercept: {model.intercept_}")
```
